```python
import json

with open("gemma4b/tokenizer.json", "r") as f:
  tokenizer_json = json.loads(f.read())
  added_tokens = tokenizer_json['added_tokens']

len(added_tokens)
```

```python
special_tokens = {}
for token in added_tokens:
  if token['content'].startswith("<unused") or token['content'].startswith("\t\t\t") or token['content'].startswith("▁▁▁▁") or token['content'].startswith("\n\n\n"):
    continue
  special_tokens[token['content']] = token['id']

len(special_tokens)
```
